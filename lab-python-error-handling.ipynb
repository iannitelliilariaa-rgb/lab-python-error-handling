Inicializar el inventorio 

````python
# Paso 1. List
products = ["t-shirt", "mug", "hat", "keychain", "books"]

# Función para inicializar el inventario
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_input = False
        while not valid_input:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity >= 0:
                    inventory[product] = quantity
                    valid_input = True
                else:
                    print("Quantity cannot be negative. Please enter a valid quantity.")
            except ValueError:
                print("Invalid input. Please enter a valid quantity.")
    return inventory

customer_inventory = initialize_inventory(products)
````

Calcular el precio total

````python
# Calcular el precio 

def calculate_total_price(customer_orders):
    total_price = 0
    for product in customer_orders:
        valid_input = False
        while not valid_input:
            try:
                # Pedimos el precio del producto actual 
                price = float(input(f"Enter the price for {product}: "))
                if price >= 0:
                    total_price += price
                    valid_input = True
                else:
                    print("Price cannot be negative. Please enter a valid price.")
            except ValueError:
                print("Invalid input. Please enter a valid price.")
    return total_price
````

Obtener los pedidos del cliente

````python
#Paso 3. Pedido del cliente
def get_customer_orders(inventory):
    customer_orders = set()
    
    # Preguntar cuántos productos quiere comprar con validación
    valid_num_orders = False
    while not valid_num_orders:
        try:
            num_orders = int(input("Enter the number of products you want to order: "))
            if num_orders >= 0:
                valid_num_orders = True
            else:
                print("Number of orders cannot be negative. Please enter a valid number.")
        except ValueError:
            print("Invalid input. Please enter a valid number.")
            
    # Pedir cada producto uno a uno y validar stock 
    for _ in range(num_orders):
        valid_product = False
        while not valid_product:
            product_name = input("Enter the name of the product you want to order: ")
            
            # Validación A: ¿El producto existe en nuestra tienda?
            if product_name not in inventory:
                print(f"Error: '{product_name}' is not in the inventory. Try again.")
                continue
                
            # Validación B: ¿Quedan unidades físicas de ese producto?
            if inventory[product_name] <= 0:
                print(f"Error: '{product_name}' is out of stock. Try again.")
                continue
                
            # Si pasa ambas pruebas, lo guardamos en el pedido
            customer_orders.add(product_name)
            valid_product = True
            
    return customer_orders
````
